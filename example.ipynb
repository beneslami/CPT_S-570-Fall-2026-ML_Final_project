```python
import ccxt
```

```python
def run_hyperliquid_ccxt_scanner():
    try:
        # 1. Initialize the built-in Hyperliquid integration
        exchange = ccxt.hyperliquid({
            'enableRateLimit': True,
            'options': {
                'defaultType': 'swap'  # Sets targeting behavior to perpetual markets
            }
        })
        
        print("⚡ Connecting to Hyperliquid standard nodes...")
        exchange.load_markets()
        
        # 2. Map human-readable names to standard CCXT derivative syntax
        asset_targets = {
            "Bitcoin": "BTC/USDC:USDC",
            "Ethereum": "ETH/USDC:USDC",
            "Solana": "SOL/USDC:USDC",
            "Gold (PAXG)": "PAXG/USDC:USDC",
        }
        
        print("\n--- 📈 Live Hyperliquid Perpetual Prices (via Unified CCXT) ---")
        
        for name, symbol in asset_targets.items():
            if symbol in exchange.symbols:
                # Fetch unified market data payload
                ticker = exchange.fetch_ticker(symbol)
                last_price = float(ticker['last'])
                
                # Check for forex naming conventions to handle decimal strings cleanly
                decimals = 5 if "uEUR" in symbol or "uGBP" in symbol or "uAUD" in symbol or "uJPY" in symbol else 2
                print(f"[{name}] {symbol} -> Price: ${last_price:,.{decimals}f}")
            else:
                print(f"❌ Symbol '{symbol}' not found on exchange.")
                
    except Exception as e:
        print(f"Execution Error: {e}")
```

```python
run_hyperliquid_ccxt_scanner()
```
