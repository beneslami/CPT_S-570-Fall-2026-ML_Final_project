```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import sklearn as skl
from pathlib import Path
import os
```

```python
path = Path("..") / "datapoints"
market_type = [
    "XAU",
    "USD",
    "QQQ.US",
    "MSFT.US",
    "HPE.US",
    "GOOGL.US",
    "GBP",
    "EUR",
    "ETH",
    "E_SANDP-500",
    "E_NQ-100",
    "E_D&J-IND",
    "BTC",
    "AMZN.US",
    "AMD.US",
    "AAPL.US"
]
timeframe = [
    "4h",
    "1d",
    "30m",
    "15m",
    "5m",
    "1m"
]
market_data = {}
for market in market_type:
    for tf in timeframe:
        file_path = path / f"{market}" / f"{market}_{tf}.csv"
        
        if not file_path.exists():
            print(f"Missing file: {file_path}")
            continue
        
        df = pd.read_csv(file_path)
        #convert time column to datetime
        df["time"] = pd.to_datetime(df["time"], errors="coerce")
        #sort by date
        df = df.sort_values(by='time').reset_index(drop=True)
        market_data[(f"{market}", f"{tf}")] = df
        print(f"Loaded {market},{tf} data with shape: {df.shape}")
```

```python
for tf in timeframe:
    if ("BTC", tf) in market_data:
        print(market_data[("BTC", tf)].head())
```

```python
for (market, tf), df in market_data.items():
    print(f"\n{market} - {tf}")
    print(df.isna().sum())
```

```python
expected_intervals = {
    "1m": pd.Timedelta(minutes=1),
    "5m": pd.Timedelta(minutes=5),
    "15m": pd.Timedelta(minutes=15),
    "30m": pd.Timedelta(minutes=30),
    "4h": pd.Timedelta(hours=4),
}

for (market, tf), df in market_data.items():
    df["Time_Gap"] = df["time"].diff()

    df["Large_Time_Gap"] = (
        df["Time_Gap"] > expected_intervals[tf]
    )
```

```python
for key, df in market_data.items():
    df["Duplicate_Date"] = df["time"].duplicated(keep=False)

for key, df in market_data.items():
    df["Return"] = df["close"].pct_change(fill_method=None)

for key, df in market_data.items():
    df["Price_Jump"] = df["Return"].abs() > 0.10
```

```python
for (market,tf), df in market_data.items():

    flagged = df[df["Large_Time_Gap"]==True]

    if not flagged.empty:
        print(f"\n{market}_{tf}")
        print(flagged[["time", "Time_Gap"]])
```
